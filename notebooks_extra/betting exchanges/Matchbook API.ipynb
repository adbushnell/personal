```python
metadata = []
pricedata = []

runners = next(market['runners'] for market in response.json()['markets'] if market['name'] == 'Winner')

for record in runners:
    
    keydata = {k: v for k, v in record.items() if k != 'prices'}
    prices = pd.DataFrame(record['prices'])
    prices['id'] = keydata['id']
    
    metadata.append(keydata)
    pricedata.append(prices)
    
runner_prices = pd.DataFrame(metadata).merge(pd.concat(pricedata), how='left', on='id')
runner_prices
```
